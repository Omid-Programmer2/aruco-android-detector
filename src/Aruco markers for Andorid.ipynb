# تولید شبکه مارکرهای ArUco
این نوت‌بوک یک تصویر شامل ۲۴ مارکر ArUco (دیکشنری 4x4) تولید می‌کند که برای تست برنامه اندروید ایده‌آل است.

```python
import cv2
import cv2.aruco as aruco
import numpy as np
import matplotlib.pyplot as plt

# ۱. تنظیمات دیکشنری
aruco_dict = aruco.getPredefinedDictionary(aruco.DICT_4X4_100)

# ۲. تنظیمات شبکه
rows, cols = 4, 6
marker_size = 200
margin = 20

# ایجاد بوم خالی سفید
sheet_width = cols * (marker_size + margin) + margin
sheet_height = rows * (marker_size + margin) + margin
sheet = np.ones((sheet_height, sheet_width), dtype=np.uint8) * 255

# ۳. تولید و جایگذاری مارکرها
for r in range(rows):
    for c in range(cols):
        marker_id = r * cols + c
        marker_img = aruco.generateImageMarker(aruco_dict, marker_id, marker_size)
        
        y = r * (marker_size + margin) + margin
        x = c * (marker_size + margin) + margin
        sheet[y:y+marker_size, x:x+marker_size] = marker_img

# ۴. ذخیره و نمایش
cv2.imwrite("aruco_marker_sheet.png", sheet)

plt.figure(figsize=(12, 8))
plt.imshow(sheet, cmap='gray')
plt.title("ArUco 4x4 Marker Sheet (IDs 0-23)")
plt.axis("off")
plt.show()

print("فایل aruco_marker_sheet.png با موفقیت ساخته شد.")
```
